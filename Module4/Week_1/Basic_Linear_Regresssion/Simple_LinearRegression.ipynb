```python
import numpy as np
```

```python
class LinearRegression():
    def __init__(self, learning_rate=0.1):
        self.w = 0
        self.b = 0
        self.learning_rate = learning_rate

    def prediction(self, x):
        return self.w * x + self.b

    def loss_function(self, y_hat, y):
        return (y_hat - y) ** 2

    def derivative_loss(self, x, y_hat, y):
        grad_w = 2 * x * (y_hat - y)
        grad_b = 2 * (y_hat - y)
        return grad_w, grad_b

    def update_parameters(self, x, y):
        y_hat = self.prediction(x)
        grad_w, grad_b = self.derivative_loss(x, y_hat, y)
        self.w -= self.learning_rate * grad_w
        self.b -= self.learning_rate * grad_b

    def train(self, X, Y, epochs=100):
        for epoch in range(epochs):
            for x, y in zip(X, Y):
                self.update_parameters(x, y)
            # Tính loss sau khi hoàn thành epoch
            total_loss = 0
            for x, y in zip(X, Y):
                y_hat = self.prediction(x)
                total_loss += self.loss_function(y_hat, y)
            avg_loss = total_loss / len(X)
            if epoch % 10 == 0:
                print(
                    f"Epoch {epoch}, "
                    f"Loss={avg_loss}, "
                    f"w={self.w}, "
                    f"b={self.b}"
                )

    def get_params(self):
        return self.w, self.b
```

```python
X = np.array([3, 4, 5, 6], dtype=float)
Y = np.array([60, 55, 66, 93], dtype=float)

mean_x = X.mean()
std_x = X.std()

mean_y = Y.mean()
std_y = Y.std()

X_norm = (X - mean_x) / std_x
Y_norm = (Y - mean_y) / std_y

model = LinearRegression(learning_rate=0.1)
model.train(X_norm, Y_norm, epochs=100)

print("Final parameters:", model.get_params())

x_new = 7
x_new_scaled = (x_new - mean_x) / std_x
y_pred = model.prediction(x_new_scaled) * std_y + mean_y

print(f"Prediction for x={x_new}: {y_pred}")
```
