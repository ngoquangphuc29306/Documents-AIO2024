```python
%cd /content/TorchJIT
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchgeometry as tgm
from PIL import Image, ImageDraw
from torchvision import transforms

from network_condition_generator import (
    ConditionGenerator,
    load_condition_generator_checkpoint,
)
from network_generator import SPADEGenerator, load_generator_checkpoint
from os import path
import matplotlib.pyplot as plt
from utils import OPT
import json
import warnings
import copy

ROOT = path.abspath('')
ROOT = path.join(ROOT, 'VITON-HD_ori/test/')

opt = OPT()
tmp = None

def remove_overlap(seg_out, warped_cm):

    assert len(warped_cm.shape) == 4

    warped_cm = warped_cm - \
        (torch.cat([seg_out[:, 1:3, :, :], seg_out[:, 5:, :, :]], dim=1)).sum(
            dim=1, keepdim=True) * warped_cm
    return warped_cm


def make_grid(N, iH, iW, opt):
    grid_x = torch.linspace(-1.0, 1.0, iW).view(1, 1,
                                                iW, 1).expand(N, iH, -1, -1)
    grid_y = torch.linspace(-1.0, 1.0, iH).view(1, iH,
                                                1, 1).expand(N, -1, iW, -1)
    if opt.cuda:
        grid = torch.cat([grid_x, grid_y], 3).cuda()
    else:
        grid = torch.cat([grid_x, grid_y], 3)
    return grid


class HRVITONModel:
    def __init__(self, condition_generator_weight_path: str, generator_weight_path: str, opt: OPT):
        self.condition_generator_weight_path = condition_generator_weight_path
        self.generator_weight_path = generator_weight_path

        self.input1_nc = 4  # cloth + cloth-mask
        self.input2_nc = opt.semantic_nc + 3  # parse_agnostic + densepose

        self.fine_width = opt.fine_width
        self.fine_height = opt.fine_height
        self.semantic_nc = opt.semantic_nc

        self.normalize_transform = transforms.Compose([
            transforms.ToTensor(),
            transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
        ])

        self._build_model(opt)

    def _build_model(self, opt: OPT):
        # self.condition_generator = ConditionGenerator(opt,
        #                                               input1_nc=self.input1_nc,
        #                                               input2_nc=self.input2_nc,
        #                                               output_nc=opt.output_nc,
        #                                               ngf=96,
        #                                               norm_layer=nn.BatchNorm2d)
        # scripted_condition_generator = torch.jit.script(self.condition_generator)
        # scripted_condition_generator.save("scripted_condition_gen.pth")
        self.condition_generator = torch.jit.load("scripted_condition_gen.pth")
        load_condition_generator_checkpoint(self.condition_generator, self.condition_generator_weight_path, opt)

        self.generator = SPADEGenerator(opt, 3+3+3)
        # global tmp
        # tmp = self.generator
        # print(self.generator.state_dict().keys())

        load_generator_checkpoint(self.generator, self.generator_weight_path, opt)
        # scripted_generator = torch.jit.script(self.generator)
        # scripted_generator.save("scripted_gen.pth")

    def get_agnostic(self, im, im_parse, pose_data):
        parse_array = np.array(im_parse)
        parse_head = ((parse_array == 4).astype(np.float32) +
                      (parse_array == 13).astype(np.float32))
        parse_lower = ((parse_array == 9).astype(np.float32) +
                       (parse_array == 12).astype(np.float32) +
                       (parse_array == 16).astype(np.float32) +
                       (parse_array == 17).astype(np.float32) +
                       (parse_array == 18).astype(np.float32) +
                       (parse_array == 19).astype(np.float32))

        agnostic = im.copy()
        agnostic_draw = ImageDraw.Draw(agnostic)

        length_a = np.linalg.norm(pose_data[5] - pose_data[2])
        length_b = np.linalg.norm(pose_data[12] - pose_data[9])
        point = (pose_data[9] + pose_data[12]) / 2
        pose_data[9] = point + (pose_data[9] - point) / length_b * length_a
        pose_data[12] = point + (pose_data[12] - point) / length_b * length_a

        r = int(length_a / 16) + 1

        # mask torso
        for i in [9, 12]:
            pointx, pointy = pose_data[i]
            agnostic_draw.ellipse(
                (pointx-r*3, pointy-r*6, pointx+r*3, pointy+r*6), 'gray', 'gray')
        agnostic_draw.line([tuple(pose_data[i])
                           for i in [2, 9]], 'gray', width=r*6)
        agnostic_draw.line([tuple(pose_data[i])
                           for i in [5, 12]], 'gray', width=r*6)
        agnostic_draw.line([tuple(pose_data[i])
                           for i in [9, 12]], 'gray', width=r*12)
        agnostic_draw.polygon([tuple(pose_data[i])
                              for i in [2, 5, 12, 9]], 'gray', 'gray')

        # mask neck
        pointx, pointy = pose_data[1]
        agnostic_draw.rectangle(
            (pointx-r*5, pointy-r*9, pointx+r*5, pointy), 'gray', 'gray')

        # mask arms
        agnostic_draw.line([tuple(pose_data[i])
                           for i in [2, 5]], 'gray', width=r*12)
        for i in [2, 5]:
            pointx, pointy = pose_data[i]
            agnostic_draw.ellipse(
                (pointx-r*5, pointy-r*6, pointx+r*5, pointy+r*6), 'gray', 'gray')
        for i in [3, 4, 6, 7]:
            if (pose_data[i-1, 0] == 0.0 and pose_data[i-1, 1] == 0.0) or (pose_data[i, 0] == 0.0 and pose_data[i, 1] == 0.0):
                continue
            agnostic_draw.line([tuple(pose_data[j])
                               for j in [i - 1, i]], 'gray', width=r*10)
            pointx, pointy = pose_data[i]
            agnostic_draw.ellipse(
                (pointx-r*5, pointy-r*5, pointx+r*5, pointy+r*5), 'gray', 'gray')

        for parse_id, pose_ids in [(14, [5, 6, 7]), (15, [2, 3, 4])]:
            mask_arm = Image.new('L', (768, 1024), 'white')
            mask_arm_draw = ImageDraw.Draw(mask_arm)
            pointx, pointy = pose_data[pose_ids[0]]
            mask_arm_draw.ellipse(
                (pointx-r*5, pointy-r*6, pointx+r*5, pointy+r*6), 'black', 'black')
            for i in pose_ids[1:]:
                if (pose_data[i-1, 0] == 0.0 and pose_data[i-1, 1] == 0.0) or (pose_data[i, 0] == 0.0 and pose_data[i, 1] == 0.0):
                    continue
                mask_arm_draw.line([tuple(pose_data[j])
                                   for j in [i - 1, i]], 'black', width=r*10)
                pointx, pointy = pose_data[i]
                if i != pose_ids[-1]:
                    mask_arm_draw.ellipse(
                        (pointx-r*5, pointy-r*5, pointx+r*5, pointy+r*5), 'black', 'black')
            mask_arm_draw.ellipse(
                (pointx-r*4, pointy-r*4, pointx+r*4, pointy+r*4), 'black', 'black')

            parse_arm = (np.array(mask_arm) / 255) * \
                (parse_array == parse_id).astype(np.float32)
            agnostic.paste(im, None, Image.fromarray(
                np.uint8(parse_arm * 255), 'L'))

        agnostic.paste(im, None, Image.fromarray(
            np.uint8(parse_head * 255), 'L'))
        agnostic.paste(im, None, Image.fromarray(
            np.uint8(parse_lower * 255), 'L'))
        return agnostic

    def _preprocess(
        self,
        person,
        cloth,
        cloth_mask,
        parse_map,
        parse_agnostic_map,
        densepose_map,
        pose_keypoints
    ):
        processed_person = transforms.Resize(
            self.fine_width, interpolation=2)(person)
        processed_person = self.normalize_transform(processed_person)

        processed_cloth = transforms.Resize(
            self.fine_width, interpolation=2)(cloth)
        processed_cloth = self.normalize_transform(processed_cloth)

        processed_densepose_map = transforms.Resize(
            self.fine_width, interpolation=2)(densepose_map)
        processed_densepose_map = self.normalize_transform(
            processed_densepose_map)  # [-1,1]

        agnostic = self.get_agnostic(person, parse_map, pose_keypoints)
        agnostic = transforms.Resize(
            self.fine_width, interpolation=2)(agnostic)
        agnostic = self.normalize_transform(agnostic)

        processed_cloth_mask = transforms.Resize(
            self.fine_width, interpolation=2)(cloth_mask)
        processed_cloth_mask = np.array(processed_cloth_mask)
        processed_cloth_mask = (processed_cloth_mask >= 128).astype(np.float32)
        processed_cloth_mask = torch.from_numpy(processed_cloth_mask)  # [0,1]
        processed_cloth_mask = processed_cloth_mask.unsqueeze(dim=0)

        processed_parse_agnostic_map = transforms.Resize(
            self.fine_width, interpolation=0)(parse_agnostic_map)
        processed_parse_agnostic_map = torch.from_numpy(
            np.array(processed_parse_agnostic_map)[None]).long()

        labels = {
            0:  ['background',  [0, 10]],
            1:  ['hair',        [1, 2]],
            2:  ['face',        [4, 13]],
            3:  ['upper',       [5, 6, 7]],
            4:  ['bottom',      [9, 12]],
            5:  ['left_arm',    [14]],
            6:  ['right_arm',   [15]],
            7:  ['left_leg',    [16]],
            8:  ['right_leg',   [17]],
            9:  ['left_shoe',   [18]],
            10: ['right_shoe',  [19]],
            11: ['socks',       [8]],
            12: ['noise',       [3, 11]]
        }

        parse_agnostic_map = torch.FloatTensor(
            20, self.fine_height, self.fine_width).zero_()
        parse_agnostic_map = parse_agnostic_map.scatter_(
            0, processed_parse_agnostic_map, 1.0)
        new_parse_agnostic_map = torch.FloatTensor(
            self.semantic_nc, self.fine_height, self.fine_width).zero_()

        for i in range(len(labels)):
            for label in labels[i][1]:
                new_parse_agnostic_map[i] += parse_agnostic_map[label]

        return {
            "person": processed_person,
            "cloth": processed_cloth,
            "cloth_mask": processed_cloth_mask,
            "parse_agnostic_map": new_parse_agnostic_map,
            "agnostic": agnostic,
            "densepose_map": processed_densepose_map
        }

    def inference(
        self,
        person: Image.Image,
        cloth: Image.Image,
        cloth_mask: Image.Image,
        parse_map: Image.Image,
        parse_agnostic_map: Image.Image,
        densepose_map: Image.Image,
        pose_keypoints: np.ndarray
    ):
        processed_data = self._preprocess(
            person, cloth, cloth_mask, parse_map, parse_agnostic_map, densepose_map, pose_keypoints)

        if opt.cuda:
            processed_person = processed_data["person"].cuda().unsqueeze(dim=0)
            processed_cloth = processed_data["cloth"].cuda().unsqueeze(dim=0)
            processed_cloth_mask = processed_data["cloth_mask"].cuda(
            ).unsqueeze(dim=0)
            processed_parse_agnostic_map = processed_data["parse_agnostic_map"].cuda(
            ).unsqueeze(dim=0)
            processed_densepose_map = processed_data["densepose_map"].cuda(
            ).unsqueeze(dim=0)
            processed_agnostic = processed_data["agnostic"].cuda(
            ).unsqueeze(dim=0)
            processed_cloth_mask = torch.FloatTensor(
                (processed_cloth_mask.detach().cpu().numpy() > 0.5).astype(np.float64)).cuda()
            processed_cloth_mask_down = F.interpolate(
                processed_cloth_mask, size=(256, 192), mode='nearest').cuda()
            processed_parse_agnostic_map_down = F.interpolate(
                processed_parse_agnostic_map, size=(256, 192), mode='nearest').cuda()
            processed_cloth_down = F.interpolate(
                processed_cloth, size=(256, 192), mode='bilinear').cuda()
            processed_densepose_map_down = F.interpolate(
                processed_densepose_map, size=(256, 192), mode='bilinear').cuda()
        else:
            processed_person = processed_data["person"].unsqueeze(dim=0)
            processed_cloth = processed_data["cloth"].unsqueeze(dim=0)
            processed_cloth_mask = processed_data["cloth_mask"].unsqueeze(
                dim=0)
            processed_parse_agnostic_map = processed_data["parse_agnostic_map"].unsqueeze(
                dim=0)
            processed_densepose_map = processed_data["densepose_map"].unsqueeze(
                dim=0)
            processed_agnostic = processed_data["agnostic"].unsqueeze(dim=0)
            processed_cloth_mask = torch.FloatTensor(
                (processed_cloth_mask.detach().cpu().numpy() > 0.5).astype(np.float64))

            processed_cloth_mask_down = F.interpolate(
                processed_cloth_mask, size=(256, 192), mode='nearest')
            processed_parse_agnostic_map_down = F.interpolate(
                processed_parse_agnostic_map, size=(256, 192), mode='nearest')
            processed_cloth_down = F.interpolate(
                processed_cloth, size=(256, 192), mode='bilinear')
            processed_densepose_map_down = F.interpolate(
                processed_densepose_map, size=(256, 192), mode='bilinear')

        input1 = torch.cat(
            [processed_cloth_down, processed_cloth_mask_down], 1)
        input2 = torch.cat(
            [processed_parse_agnostic_map_down, processed_densepose_map_down], 1)

        # print(input1.is_cuda, input2.is_cuda)
        # torch.save(input1, "input1.pt")
        # torch.save(input2, "input2.pt")

        flow_list, fake_segmap, warped_cloth_paired, warped_clothmask_paired = self.condition_generator(
            opt, input1, input2)

        cloth_mask = torch.ones_like(fake_segmap)
        cloth_mask[:, 3:4, :, :] = warped_clothmask_paired
        fake_segmap = fake_segmap * cloth_mask

        gauss = tgm.image.GaussianBlur((15, 15), (3, 3))
        gauss = gauss.cuda()
        fake_parse_gauss = gauss(F.interpolate(fake_segmap, size=(
            opt.fine_height, opt.fine_width), mode='bilinear'))

        fake_parse_gauss = gauss(F.interpolate(fake_segmap, size=(
            opt.fine_height, opt.fine_width), mode='bilinear'))
        fake_parse = fake_parse_gauss.argmax(dim=1)[:, None]

        if opt.cuda:
            old_parse = torch.FloatTensor(fake_parse.size(
                0), 13, opt.fine_height, opt.fine_width).zero_().cuda()
        else:
            old_parse = torch.FloatTensor(fake_parse.size(
                0), 13, opt.fine_height, opt.fine_width).zero_()
        old_parse.scatter_(1, fake_parse, 1.0)

        labels = {
            0:  ['background',  [0]],
            1:  ['paste',       [2, 4, 7, 8, 9, 10, 11]],
            2:  ['upper',       [3]],
            3:  ['hair',        [1]],
            4:  ['left_arm',    [5]],
            5:  ['right_arm',   [6]],
            6:  ['noise',       [12]]
        }
        if opt.cuda:
            parse = torch.FloatTensor(fake_parse.size(
                0), 7, opt.fine_height, opt.fine_width).zero_().cuda()
        else:
            parse = torch.FloatTensor(fake_parse.size(
                0), 7, opt.fine_height, opt.fine_width).zero_()
        for i in range(len(labels)):
            for label in labels[i][1]:
                parse[:, i] += old_parse[:, label]

                # warped cloth
        N, _, iH, iW = processed_cloth.shape
        flow = F.interpolate(
            flow_list[-1].permute(0, 3, 1, 2), size=(iH, iW), mode='bilinear').permute(0, 2, 3, 1)
        flow_norm = torch.cat(
            [flow[:, :, :, 0:1] / ((96 - 1.0) / 2.0), flow[:, :, :, 1:2] / ((128 - 1.0) / 2.0)], 3)
        with warnings.catch_warnings():
          grid = make_grid(N, iH, iW, opt)
          warped_grid = grid + flow_norm
          warped_cloth = F.grid_sample(
              processed_cloth, warped_grid, padding_mode='border', align_corners=False)
          warped_clothmask = F.grid_sample(
              processed_cloth_mask, warped_grid, padding_mode='border', align_corners=False)

        warped_clothmask = remove_overlap(
            F.softmax(fake_parse_gauss, dim=1), warped_clothmask)
        warped_cloth = warped_cloth * warped_clothmask + \
            torch.ones_like(warped_cloth) * (1-warped_clothmask)
        global tmp
        tmp = (torch.cat(
            (processed_agnostic, processed_densepose_map, warped_cloth), dim=1).clone(), parse.clone())
        output = self.generator(torch.cat(
            (processed_agnostic, processed_densepose_map, warped_cloth), dim=1), parse)
        # traced_generator = torch.jit.trace(self.generator, (torch.cat(
        #     (processed_agnostic, processed_densepose_map, warped_cloth), dim=1), parse))
        # traced_generator.save("traced_gen.pth")
        # scripted_gen = torch.jit.script(self.generator)
        # scripted_gen.save("scripted_gen.pth")
        # # import ipdb; ipdb.set_trace()
        denormalized_warped_cloth = ((output.cpu().detach().numpy(
        ) / 2 + 0.5)*255).astype(np.uint8).squeeze().transpose(1, 2, 0)
        return denormalized_warped_cloth

model_2 = HRVITONModel(
    condition_generator_weight_path="./condition_gen.pth",
    generator_weight_path="gen.pth",
    opt=opt
)

person = Image.open(path.join(ROOT, "image/03797_00.jpg"))
cloth = Image.open(path.join(ROOT, "cloth/01171_00.jpg")).convert('RGB')
cloth_mask = Image.open(path.join(ROOT, "cloth-mask/01171_00.jpg"))
person_parse_agnostic_map = Image.open(
    path.join(ROOT, "image-parse-agnostic-v3.2/03797_00.png"))
person_parse_map = Image.open(
    path.join(ROOT, "image-parse-v3/03797_00.png"))
densepose_map = Image.open(path.join(ROOT, "image-densepose/03797_00.jpg"))
# import ipdb
# ipdb.set_trace()

with open(path.join(ROOT, "openpose_json/03797_00_keypoints.json")) as f:
    pose_label = json.load(f)
    pose_data = pose_label['people'][0]['pose_keypoints_2d']
    pose_data = np.array(pose_data)
    pose_data = pose_data.reshape((-1, 3))[:, :2]

out = model_2.inference(person, cloth, cloth_mask, person_parse_map,
                        person_parse_agnostic_map, densepose_map, pose_data)
out = Image.fromarray(out)
images = {'person': person,
          'cloth': cloth,
          'cloth_mask': cloth_mask,
          'person_parse_agnostic_map': person_parse_agnostic_map,
          'person_parse_map': person_parse_map,
          'densepose_map': densepose_map,
          'out': out}

fig, axes = plt.subplots(3, 3, figsize=(10, 10))
for i, (key, value) in enumerate(images.items()):
    ax = axes[i//3, i % 3]
    ax.imshow(value)
    ax.set_title(key)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchgeometry as tgm
from PIL import Image, ImageDraw
from torchvision import transforms

from network_condition_generator import (
    ConditionGenerator,
    load_condition_generator_checkpoint,
)
from network_generator_2 import SPADEGenerator, load_generator_checkpoint
from os import path
import matplotlib.pyplot as plt
from utils import OPT
import warnings
import json

ROOT = path.abspath('')
ROOT = path.join(ROOT, 'VITON-HD_ori/test/')

opt = OPT()


def remove_overlap(seg_out, warped_cm):

    assert len(warped_cm.shape) == 4

    warped_cm = warped_cm - \
        (torch.cat([seg_out[:, 1:3, :, :], seg_out[:, 5:, :, :]], dim=1)).sum(
            dim=1, keepdim=True) * warped_cm
    return warped_cm


def make_grid(N, iH, iW, opt):
    grid_x = torch.linspace(-1.0, 1.0, iW).view(1, 1,
                                                iW, 1).expand(N, iH, -1, -1)
    grid_y = torch.linspace(-1.0, 1.0, iH).view(1, iH,
                                                1, 1).expand(N, -1, iW, -1)
    if opt.cuda:
        grid = torch.cat([grid_x, grid_y], 3).cuda()
    else:
        grid = torch.cat([grid_x, grid_y], 3)
    return grid


class HRVITONModel:
    def __init__(self, condition_generator_weight_path: str, generator_weight_path: str, opt: OPT):
        self.condition_generator_weight_path = condition_generator_weight_path
        self.generator_weight_path = generator_weight_path

        self.input1_nc = 4  # cloth + cloth-mask
        self.input2_nc = opt.semantic_nc + 3  # parse_agnostic + densepose

        self.fine_width = opt.fine_width
        self.fine_height = opt.fine_height
        self.semantic_nc = opt.semantic_nc

        self.normalize_transform = transforms.Compose([
            transforms.ToTensor(),
            transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
        ])

        self._build_model(opt)

    def _build_model(self, opt: OPT):
        self.condition_generator = ConditionGenerator(opt,
                                                      input1_nc=self.input1_nc,
                                                      input2_nc=self.input2_nc,
                                                      output_nc=opt.output_nc,
                                                      ngf=96,
                                                      norm_layer=nn.BatchNorm2d)
        # scripted_condition_generator = torch.jit.script(self.condition_generator)
        # scripted_condition_generator.save("scripted_condition_gen.pth")
        # self.condition_generator = torch.jit.load("scripted_condition_gen.pth")
        load_condition_generator_checkpoint(
            self.condition_generator, self.condition_generator_weight_path, opt)

        self.generator = SPADEGenerator(opt, 3+3+3)
        # scripted_generator = torch.jit.script(self.generator)
        # scripted_generator.save("scripted_gen.pth")
        load_generator_checkpoint(
            self.generator, self.generator_weight_path, opt)

    def get_agnostic(self, im, im_parse, pose_data):
        parse_array = np.array(im_parse)
        parse_head = ((parse_array == 4).astype(np.float32) +
                      (parse_array == 13).astype(np.float32))
        parse_lower = ((parse_array == 9).astype(np.float32) +
                       (parse_array == 12).astype(np.float32) +
                       (parse_array == 16).astype(np.float32) +
                       (parse_array == 17).astype(np.float32) +
                       (parse_array == 18).astype(np.float32) +
                       (parse_array == 19).astype(np.float32))

        agnostic = im.copy()
        agnostic_draw = ImageDraw.Draw(agnostic)

        length_a = np.linalg.norm(pose_data[5] - pose_data[2])
        length_b = np.linalg.norm(pose_data[12] - pose_data[9])
        point = (pose_data[9] + pose_data[12]) / 2
        pose_data[9] = point + (pose_data[9] - point) / length_b * length_a
        pose_data[12] = point + (pose_data[12] - point) / length_b * length_a

        r = int(length_a / 16) + 1

        # mask torso
        for i in [9, 12]:
            pointx, pointy = pose_data[i]
            agnostic_draw.ellipse(
                (pointx-r*3, pointy-r*6, pointx+r*3, pointy+r*6), 'gray', 'gray')
        agnostic_draw.line([tuple(pose_data[i])
                           for i in [2, 9]], 'gray', width=r*6)
        agnostic_draw.line([tuple(pose_data[i])
                           for i in [5, 12]], 'gray', width=r*6)
        agnostic_draw.line([tuple(pose_data[i])
                           for i in [9, 12]], 'gray', width=r*12)
        agnostic_draw.polygon([tuple(pose_data[i])
                              for i in [2, 5, 12, 9]], 'gray', 'gray')

        # mask neck
        pointx, pointy = pose_data[1]
        agnostic_draw.rectangle(
            (pointx-r*5, pointy-r*9, pointx+r*5, pointy), 'gray', 'gray')

        # mask arms
        agnostic_draw.line([tuple(pose_data[i])
                           for i in [2, 5]], 'gray', width=r*12)
        for i in [2, 5]:
            pointx, pointy = pose_data[i]
            agnostic_draw.ellipse(
                (pointx-r*5, pointy-r*6, pointx+r*5, pointy+r*6), 'gray', 'gray')
        for i in [3, 4, 6, 7]:
            if (pose_data[i-1, 0] == 0.0 and pose_data[i-1, 1] == 0.0) or (pose_data[i, 0] == 0.0 and pose_data[i, 1] == 0.0):
                continue
            agnostic_draw.line([tuple(pose_data[j])
                               for j in [i - 1, i]], 'gray', width=r*10)
            pointx, pointy = pose_data[i]
            agnostic_draw.ellipse(
                (pointx-r*5, pointy-r*5, pointx+r*5, pointy+r*5), 'gray', 'gray')

        for parse_id, pose_ids in [(14, [5, 6, 7]), (15, [2, 3, 4])]:
            mask_arm = Image.new('L', (768, 1024), 'white')
            mask_arm_draw = ImageDraw.Draw(mask_arm)
            pointx, pointy = pose_data[pose_ids[0]]
            mask_arm_draw.ellipse(
                (pointx-r*5, pointy-r*6, pointx+r*5, pointy+r*6), 'black', 'black')
            for i in pose_ids[1:]:
                if (pose_data[i-1, 0] == 0.0 and pose_data[i-1, 1] == 0.0) or (pose_data[i, 0] == 0.0 and pose_data[i, 1] == 0.0):
                    continue
                mask_arm_draw.line([tuple(pose_data[j])
                                   for j in [i - 1, i]], 'black', width=r*10)
                pointx, pointy = pose_data[i]
                if i != pose_ids[-1]:
                    mask_arm_draw.ellipse(
                        (pointx-r*5, pointy-r*5, pointx+r*5, pointy+r*5), 'black', 'black')
            mask_arm_draw.ellipse(
                (pointx-r*4, pointy-r*4, pointx+r*4, pointy+r*4), 'black', 'black')

            parse_arm = (np.array(mask_arm) / 255) * \
                (parse_array == parse_id).astype(np.float32)
            agnostic.paste(im, None, Image.fromarray(
                np.uint8(parse_arm * 255), 'L'))

        agnostic.paste(im, None, Image.fromarray(
            np.uint8(parse_head * 255), 'L'))
        agnostic.paste(im, None, Image.fromarray(
            np.uint8(parse_lower * 255), 'L'))
        return agnostic

    def _preprocess(
        self,
        person,
        cloth,
        cloth_mask,
        parse_map,
        parse_agnostic_map,
        densepose_map,
        pose_keypoints
    ):
        processed_person = transforms.Resize(
            self.fine_width, interpolation=2)(person)
        processed_person = self.normalize_transform(processed_person)

        processed_cloth = transforms.Resize(
            self.fine_width, interpolation=2)(cloth)
        processed_cloth = self.normalize_transform(processed_cloth)

        processed_densepose_map = transforms.Resize(
            self.fine_width, interpolation=2)(densepose_map)
        processed_densepose_map = self.normalize_transform(
            processed_densepose_map)  # [-1,1]

        agnostic = self.get_agnostic(person, parse_map, pose_keypoints)
        agnostic = transforms.Resize(
            self.fine_width, interpolation=2)(agnostic)
        agnostic = self.normalize_transform(agnostic)

        processed_cloth_mask = transforms.Resize(
            self.fine_width, interpolation=2)(cloth_mask)
        processed_cloth_mask = np.array(processed_cloth_mask)
        processed_cloth_mask = (processed_cloth_mask >= 128).astype(np.float32)
        processed_cloth_mask = torch.from_numpy(processed_cloth_mask)  # [0,1]
        processed_cloth_mask = processed_cloth_mask.unsqueeze(dim=0)

        processed_parse_agnostic_map = transforms.Resize(
            self.fine_width, interpolation=0)(parse_agnostic_map)
        processed_parse_agnostic_map = torch.from_numpy(
            np.array(processed_parse_agnostic_map)[None]).long()

        labels = {
            0:  ['background',  [0, 10]],
            1:  ['hair',        [1, 2]],
            2:  ['face',        [4, 13]],
            3:  ['upper',       [5, 6, 7]],
            4:  ['bottom',      [9, 12]],
            5:  ['left_arm',    [14]],
            6:  ['right_arm',   [15]],
            7:  ['left_leg',    [16]],
            8:  ['right_leg',   [17]],
            9:  ['left_shoe',   [18]],
            10: ['right_shoe',  [19]],
            11: ['socks',       [8]],
            12: ['noise',       [3, 11]]
        }

        parse_agnostic_map = torch.FloatTensor(
            20, self.fine_height, self.fine_width).zero_()
        parse_agnostic_map = parse_agnostic_map.scatter_(
            0, processed_parse_agnostic_map, 1.0)
        new_parse_agnostic_map = torch.FloatTensor(
            self.semantic_nc, self.fine_height, self.fine_width).zero_()

        for i in range(len(labels)):
            for label in labels[i][1]:
                new_parse_agnostic_map[i] += parse_agnostic_map[label]

        return {
            "person": processed_person,
            "cloth": processed_cloth,
            "cloth_mask": processed_cloth_mask,
            "parse_agnostic_map": new_parse_agnostic_map,
            "agnostic": agnostic,
            "densepose_map": processed_densepose_map
        }

    def inference(
        self,
        person: Image.Image,
        cloth: Image.Image,
        cloth_mask: Image.Image,
        parse_map: Image.Image,
        parse_agnostic_map: Image.Image,
        densepose_map: Image.Image,
        pose_keypoints: np.ndarray
    ):
        processed_data = self._preprocess(
            person, cloth, cloth_mask, parse_map, parse_agnostic_map, densepose_map, pose_keypoints)

        if opt.cuda:
            processed_person = processed_data["person"].cuda().unsqueeze(dim=0)
            processed_cloth = processed_data["cloth"].cuda().unsqueeze(dim=0)
            processed_cloth_mask = processed_data["cloth_mask"].cuda(
            ).unsqueeze(dim=0)
            processed_parse_agnostic_map = processed_data["parse_agnostic_map"].cuda(
            ).unsqueeze(dim=0)
            processed_densepose_map = processed_data["densepose_map"].cuda(
            ).unsqueeze(dim=0)
            processed_agnostic = processed_data["agnostic"].cuda(
            ).unsqueeze(dim=0)
            processed_cloth_mask = torch.FloatTensor(
                (processed_cloth_mask.detach().cpu().numpy() > 0.5).astype(np.float64)).cuda()
            processed_cloth_mask_down = F.interpolate(
                processed_cloth_mask, size=(256, 192), mode='nearest').cuda()
            processed_parse_agnostic_map_down = F.interpolate(
                processed_parse_agnostic_map, size=(256, 192), mode='nearest').cuda()
            processed_cloth_down = F.interpolate(
                processed_cloth, size=(256, 192), mode='bilinear').cuda()
            processed_densepose_map_down = F.interpolate(
                processed_densepose_map, size=(256, 192), mode='bilinear').cuda()
        else:
            processed_person = processed_data["person"].unsqueeze(dim=0)
            processed_cloth = processed_data["cloth"].unsqueeze(dim=0)
            processed_cloth_mask = processed_data["cloth_mask"].unsqueeze(
                dim=0)
            processed_parse_agnostic_map = processed_data["parse_agnostic_map"].unsqueeze(
                dim=0)
            processed_densepose_map = processed_data["densepose_map"].unsqueeze(
                dim=0)
            processed_agnostic = processed_data["agnostic"].unsqueeze(dim=0)
            processed_cloth_mask = torch.FloatTensor(
                (processed_cloth_mask.detach().cpu().numpy() > 0.5).astype(np.float64))

            processed_cloth_mask_down = F.interpolate(
                processed_cloth_mask, size=(256, 192), mode='nearest')
            processed_parse_agnostic_map_down = F.interpolate(
                processed_parse_agnostic_map, size=(256, 192), mode='nearest')
            processed_cloth_down = F.interpolate(
                processed_cloth, size=(256, 192), mode='bilinear')
            processed_densepose_map_down = F.interpolate(
                processed_densepose_map, size=(256, 192), mode='bilinear')

        input1 = torch.cat(
            [processed_cloth_down, processed_cloth_mask_down], 1)
        input2 = torch.cat(
            [processed_parse_agnostic_map_down, processed_densepose_map_down], 1)

        flow_list, fake_segmap, warped_cloth_paired, warped_clothmask_paired = self.condition_generator(
            opt, input1, input2)

        cloth_mask = torch.ones_like(fake_segmap)
        cloth_mask[:, 3:4, :, :] = warped_clothmask_paired
        fake_segmap = fake_segmap * cloth_mask

        gauss = tgm.image.GaussianBlur((15, 15), (3, 3))
        gauss = gauss.cuda()
        fake_parse_gauss = gauss(F.interpolate(fake_segmap, size=(
            opt.fine_height, opt.fine_width), mode='bilinear'))

        fake_parse_gauss = gauss(F.interpolate(fake_segmap, size=(
            opt.fine_height, opt.fine_width), mode='bilinear'))
        fake_parse = fake_parse_gauss.argmax(dim=1)[:, None]

        if opt.cuda:
            old_parse = torch.FloatTensor(fake_parse.size(
                0), 13, opt.fine_height, opt.fine_width).zero_().cuda()
        else:
            old_parse = torch.FloatTensor(fake_parse.size(
                0), 13, opt.fine_height, opt.fine_width).zero_()
        old_parse.scatter_(1, fake_parse, 1.0)

        labels = {
            0:  ['background',  [0]],
            1:  ['paste',       [2, 4, 7, 8, 9, 10, 11]],
            2:  ['upper',       [3]],
            3:  ['hair',        [1]],
            4:  ['left_arm',    [5]],
            5:  ['right_arm',   [6]],
            6:  ['noise',       [12]]
        }
        if opt.cuda:
            parse = torch.FloatTensor(fake_parse.size(
                0), 7, opt.fine_height, opt.fine_width).zero_().cuda()
        else:
            parse = torch.FloatTensor(fake_parse.size(
                0), 7, opt.fine_height, opt.fine_width).zero_()
        for i in range(len(labels)):
            for label in labels[i][1]:
                parse[:, i] += old_parse[:, label]

                # warped cloth
        N, _, iH, iW = processed_cloth.shape
        flow = F.interpolate(
            flow_list[-1].permute(0, 3, 1, 2), size=(iH, iW), mode='bilinear').permute(0, 2, 3, 1)
        flow_norm = torch.cat(
            [flow[:, :, :, 0:1] / ((96 - 1.0) / 2.0), flow[:, :, :, 1:2] / ((128 - 1.0) / 2.0)], 3)

        with warnings.catch_warnings():
          grid = make_grid(N, iH, iW, opt)
          warped_grid = grid + flow_norm
          warped_cloth = F.grid_sample(
              processed_cloth, warped_grid, padding_mode='border', align_corners=False)
          warped_clothmask = F.grid_sample(
              processed_cloth_mask, warped_grid, padding_mode='border', align_corners=False)

        warped_clothmask = remove_overlap(
            F.softmax(fake_parse_gauss, dim=1), warped_clothmask)
        warped_cloth = warped_cloth * warped_clothmask + \
            torch.ones_like(warped_cloth) * (1-warped_clothmask)

        output = self.generator(torch.cat(
            (processed_agnostic, processed_densepose_map, warped_cloth), dim=1), parse)
        # # import ipdb; ipdb.set_trace()
        denormalized_warped_cloth = ((output.cpu().detach().numpy(
        ) / 2 + 0.5)*255).astype(np.uint8).squeeze().transpose(1, 2, 0)
        return denormalized_warped_cloth

model = HRVITONModel(
    condition_generator_weight_path="./condition_gen.pth",
    generator_weight_path="gen.pth",
    opt=opt
)

person = Image.open(path.join(ROOT, "image/03797_00.jpg"))
cloth = Image.open(path.join(ROOT, "cloth/01171_00.jpg")).convert('RGB')
cloth_mask = Image.open(path.join(ROOT, "cloth-mask/01171_00.jpg"))
person_parse_agnostic_map = Image.open(
    path.join(ROOT, "image-parse-agnostic-v3.2/03797_00.png"))
person_parse_map = Image.open(
    path.join(ROOT, "image-parse-v3/03797_00.png"))
densepose_map = Image.open(path.join(ROOT, "image-densepose/03797_00.jpg"))
# import ipdb
# ipdb.set_trace()

with open(path.join(ROOT, "openpose_json/03797_00_keypoints.json")) as f:
    pose_label = json.load(f)
    pose_data = pose_label['people'][0]['pose_keypoints_2d']
    pose_data = np.array(pose_data)
    pose_data = pose_data.reshape((-1, 3))[:, :2]

# out = model.inference(person, cloth, cloth_mask, person_parse_map,
#                         person_parse_agnostic_map, densepose_map, pose_data)
# out = Image.fromarray(out)
# images = {'person': person,
#           'cloth': cloth,
#           'cloth_mask': cloth_mask,
#           'person_parse_agnostic_map': person_parse_agnostic_map,
#           'person_parse_map': person_parse_map,
#           'densepose_map': densepose_map,
#           'out': out}

# fig, axes = plt.subplots(3, 3, figsize=(10, 10))
# for i, (key, value) in enumerate(images.items()):
#     ax = axes[i//3, i % 3]
#     ax.imshow(value)
#     ax.set_title(key)
#     ax.axis('off')
# plt.tight_layout()
# plt.show()
```
