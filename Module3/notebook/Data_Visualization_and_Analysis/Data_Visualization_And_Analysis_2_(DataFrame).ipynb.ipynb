## **Concatenate data**

```python
import pandas as pd

# Create the first DataFrame (df1)
data1 = {'col1': [1, 2], 'col2': [3, 4]}
df1 = pd.DataFrame(data1)
print("DataFrame 1:")
print(df1)

# Create the second DataFrame (df2)
data2 = {'col1': [5, 6], 'col2': [7, 8]}
df2 = pd.DataFrame(data2)
print("\nDataFrame 2:")
print(df2)

# Concatenate the two DataFrames
df3 = pd.concat([df1, df2])
df4 = df3.reset_index(drop=True)
print(df4)
```

```python
df3 = pd.concat([df1, df2], ignore_index=True)
df3
```

## **Date time**

```python
datetime = pd.to_datetime('2006-12-01 00:00:00.000 +0700', utc = True)
print(datetime)
print(datetime.year)
print(datetime.month)
print(datetime.day)
print(datetime.hour)
print(datetime.minute)
print(datetime.second)
print(datetime.tz)
```

```python
datetime = pd.to_datetime('2006-12-01 00:00:00.000', utc = False)
print(datetime)
print(datetime.year)
print(datetime.month)
print(datetime.day)
print(datetime.hour)
print(datetime.minute)
print(datetime.second)
```

```python
print(pd.Timestamp(year=2006, month=3, day=29, hour=1, minute=35, second=10))
print(pd.Timestamp('2006-03-29 01:35:10'))
print(pd.Timestamp('Sep 04, 1982 1:35:18'))
```

```python
df = pd.read_csv('/content/WeatherHistory2D.csv')
df['Formatted Date'] = pd.to_datetime(df['Formatted Date'], utc=True)

df['year'] = df['Formatted Date'].dt.year
df['month'] = df['Formatted Date'].dt.month
df['day'] = df['Formatted Date'].dt.day
df['hour'] = df['Formatted Date'].dt.hour
df.drop('Formatted Date', axis = 1, inplace=True)
df
```
