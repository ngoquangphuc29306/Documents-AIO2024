```python
data = pd.read_csv('iris.csv')
pd.plotting.parallel_coordinates(data, "species")
```

```python
sns.jointplot(x="sepal_length", y="sepal_width",
              data=data, hue="species")
plt.show()
```

```python
sns.jointplot(x="petal_length", y="petal_width",
              data=data, hue="species")
plt.show()
```

```python
sns.stripplot(y="sepal_length", x="species",
              data=data, hue="species")
plt.show()
```

```python
sns.stripplot(y="petal_length", x="species",
              data=data, hue="species")
plt.show()
```

```python
sns.boxplot(x="species", y="sepal_length", data=data, hue="species")
plt.show()
```

```python
# Correlation coefficient
sns.pairplot(data=data, hue="species")
plt.show()
```

```python
sns.heatmap(data.iloc[:,0:4].corr(),
            annot=True, fmt="f")
plt.show()
```
