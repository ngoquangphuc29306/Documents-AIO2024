## **Addition between two series**

```python
center1 = pd.Series([12, 23, 31, 11, 9],
                    index=list(['C++', 'Golang', 'Java', 'Python', 'Swwift']),
                    name='num_registered')

center2 = pd.Series([42, 18, 44, 49, 27],
                    index=list(['C++', 'Golang', 'Java', 'Python', 'Swwift']),
                    name='num_registered')

total = center1 + center2
total
```

```python
center1 = pd.Series([12, 23, 31, 11, 9],
                    index=list(['C++', 'Golang', 'Java', 'Python', 'Swwift']),
                    name='num_registered')

center2 = pd.Series([18, 44, 49, 27],
                    index=list(['Golang', 'Java', 'Python', 'Swwift']),
                    name='num_registered')

total = center1 + center2
total
```

## **Group By**

```python
import pandas as pd

data = pd.Series([4, 1, 0, 2, 2, 1],
                 index=['C++', 'Golang', 'Java', 'Golang', 'Java', 'C++'],
                 name='num_dropped')
data
```

```python
data.groupby(level=0).sum()
```

```python
data.groupby(data < 4).sum()
```

```python
print(data['C++'].sum())
```

# **Case Study**

Temperature forecasting

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/weatherHistory_v1.csv')
data_series = df['Temperature (C)']

data_series.describe()
```

```python
data_filled = data_series.interpolate()
```

```python
data_filled.plot()
```

```python
data_small = data_filled[:150]
data_small.plot.box()
```

```python
# Reduce Noise
# With Simple moving average
data_smooth_SMA = data_filled.rolling(window = 5).mean()
data_smooth_SMA[:100].plot()
```

```python
# With Exponentail moving average
data_smooth_EMA = data_filled.ewm(com=2).mean()
data_smooth_EMA[:100].plot()
```

```python
data_small.hist()
```

```python
data_small.plot.kde() # Kernel density estimation
# Biến các điểm dữ liệu rời ạc thành một đường cong mượt
```
